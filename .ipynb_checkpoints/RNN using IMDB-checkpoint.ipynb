```python
import pandas as pd
from tensorflow.keras import Sequential
from tensorflow.keras.layers import TextVectorization, Embedding, LSTM, Dense
from sklearn.model_selection import train_test_split
```

```python
# Load data
df = pd.read_csv("IMDB Dataset.csv")
x = df["review"].astype(str).values
y = df["sentiment"].map({"positive": 1, "negative": 0}).values
```

```python
# Split data
xtrain, xtest, ytrain, ytest = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# Turn text into number sequences
vectorize = TextVectorization(
    max_tokens=10000,
    output_sequence_length=200
)
vectorize.adapt(xtrain)
```

```python
# Build model
model = Sequential([
    vectorize,
    Embedding(10000, 128),
    LSTM(64),
    Dense(1, activation="sigmoid")
])
```

```python
# Train and evaluate
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
model.fit(xtrain, ytrain, epochs=5, batch_size=64, validation_split=0.2)
model.evaluate(xtest, ytest)
```

```python
# Predict one review
review = input("Enter a movie review: ")

probability = model.predict([review], verbose=0)[0][0]
sentiment = "positive" if probability >= 0.5 else "negative"
```
